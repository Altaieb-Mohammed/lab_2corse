```python
import pandas as pd

reviews = []
labels = []

with open('revews.txt', 'r', encoding='utf-8') as f:
    for line in f:
        # Предположим, что строка формата: "positive\tThis movie is great!"
        parts = line.strip().split('\t')
        if len(parts) == 2:
            label, review = parts
            labels.append(label)
            reviews.append(review)

df = pd.DataFrame({'review': reviews, 'sentiment': labels})
df.to_csv('reviews.csv', index=False)
print("CSV файл создан!")
```
